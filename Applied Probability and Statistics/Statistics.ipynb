```python
import random
import statistics
import numpy as np
samples = random.sample(range(1, 10001), 10000)
mean = sum(samples) / len(samples)
mode = statistics.mode(samples)
median = statistics.median(samples)
print("Mean:", mean)
print("Mode:", mode)
print("Median:", median)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Population
np.random.seed(42)
population = np.random.normal(loc=50, scale=10, size=10000)

# Settings
sample_size = 30
num_samples = 1000

sample_means = []
sample_vars = []

# Draw many samples
for _ in range(num_samples):
    sample = np.random.choice(population, size=sample_size)
    sample_means.append(np.mean(sample))
    sample_vars.append(np.var(sample, ddof=1))  # ddof=1 -> sample variance

# Results
print("Population mean:", np.mean(population))
print("Population variance:", np.var(population))
print("Mean of sample means:", np.mean(sample_means))
print("Mean of sample variances:", np.mean(sample_vars))

# Plot sampling distributions
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.hist(sample_means, bins=30, color="skyblue", edgecolor="black")
plt.title("Sampling Distribution of Mean")

plt.subplot(1, 2, 2)
plt.hist(sample_vars, bins=30, color="salmon", edgecolor="black")
plt.title("Sampling Distribution of Variance")

plt.tight_layout()
plt.show()
```
