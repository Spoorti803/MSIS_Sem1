```python
def birthday_probab(n):
    probab_no_match = 1.0
    for i in range(n):
        probab_no_match *= (365-i)/365
    # probab_match = 1-probab_no_match
    return probab_no_match
```

```python
n=int(input("Enter the number of values"))

probab_n = birthday_probab(n)
print(f"The probablity of {n} is",probab_n)
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
import pandas as pd

max_people = 365
people_list = []
probab_list = []

for num_p in range(1, max_people + 1):
    prob_no_match_current = birthday_probab(num_p)
    prob_match_current = 1 - prob_no_match_current
    people_list.append(num_p)
    probab_list.append(prob_match_current)

df = pd.DataFrame({
    'No of people': people_list,
    'probablity': probab_list
})

plt.xlabel('No of people')
plt.ylabel('Probablity')
plt.title('Birthday Problem')
plt.plot(df['No of people'],df['probablity'])
plt.show()
```

```python
'''    Addition rule for disjoint events

A die is rolled 10,000 times.

Define: A=getting 1, B= getting 6, check the events are joint or disjoint

Verify: P(A∪B)=P(A)+P(B)
'''
import random
n=10000
results = [random.randint(1,6) for _ in range(n)]

A = results.count(1)
B = results.count(6)

A_and_B = A + B

probab_A = A/n
probab_B = B/n
probab_A_union_B = A_and_B/n

print(f"P(A) = {probab_A}")
print(f"P(B) = {probab_B}")
print(f"P(A ∪ B) = {probab_A_union_B}")

print("Number of rolls:", 10000)
print("A (getting 1):", A)
print("B (getting 6):", B)

print("\nP(A) =", probab_A)
print("P(B) =", probab_B)
print("P(A ∪ B) =", probab_A_union_B)

print("\nP(A) + P(B) =", probab_A + probab_B)

if(probab_A_union_B == probab_A + probab_B):
    print("The events are disjoint.")
```

```python
'''    Addition rule for overlapping events

For one die roll, define: A=getting an even number A={2,4,6} and B=getting a number greater than 3 B={4,5,6}

Verify: P(A∪B)=P(A)+P(B)−P(A∩B)
'''
import random
n=10000
results = [random.randint(1,6) for _ in range(n)]

A = sum(1 for i in results if i in [2,4,6] )
B = sum(1 for i in results if i in [4,5,6] )
A_and_B = sum(1 for i in results if i in [4,6])
A_or_B = sum(1 for i in results if i in [2,4,5,6])

probab_A = A/n
probab_B = B/n
probab_A_union_B = A_or_B/n
probab_A_intersect_B = A_and_B/n

print("\nP(A) =", probab_A)
print("P(B) =", probab_B)
print("P(A ∩ B) =", probab_A_intersect_B)
print("P(A ∪ B) =", probab_A_union_B)

if (probab_A_union_B == probab_A + probab_B - probab_A_intersect_B):
    print("The events are disjoint.")
else:
    print("The events are not disjoint.")
```

```python
'''    Multiplication rule for independent events

A coin is tossed and a die is rolled 10,000 times.

Define:

    obtaining Head

    obtaining an even number

Verify: P(A∩B)=P(A)P(B)
'''
import random
n = 10000

# Counters
heads = 0
even = 0
head_and_even = 0

for _ in range(n):
    coin = random.choice(["Head", "Tail"])
    die = random.randint(1, 6)

    if coin == "Head":
        heads += 1

    if die % 2 == 0:
        even += 1

    if coin == "Head" and die % 2 == 0:
        head_and_even += 1


P_A = heads / n
P_B = even / n
P_A_intersection_B = head_and_even / n

P_A_times_P_B = P_A * P_B



print("\nNumber of Heads:", heads)
print("Number of Even results:", even)
print("Number of Head AND Even results:", head_and_even)

print("\nP(A) =", P_A)
print("P(B) =", P_B)
print("P(A ∩ B) =", P_A_intersection_B)

print("\nP(A) × P(B) =", P_A_times_P_B)

if abs(P_A_intersection_B - P_A_times_P_B) < 0.01:
    print("\nMultiplication rule verified!")
else:
    print("\nValues are approximately equal.")
```
