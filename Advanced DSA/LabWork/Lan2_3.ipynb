Implementing Singly-linked list

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None  # Initialize next as None

class SinglyLinkedList:
    def __init__(self):
        self.head = None  # Initialize head of the list as None

    # Method to print the linked list
    def print_list(self):
        current = self.head
        elements = []
        while current:
            elements.append(current.data)
            current = current.next
        print(" -> ".join(map(str, elements)) if elements else "List is empty")

    # Method to insert a new node at the beginning
    def insert_at_beginning(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    # Method to insert a new node at the end
    def insert_at_end(self, data):
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            return
        last = self.head
        while last.next:
            last = last.next
        last.next = new_node

    # Method to insert a new node after a given node's data
    def insert_after(self, prev_node_data, data):
        if prev_node_data is None:
            print("Previous node data cannot be None.")
            return

        current = self.head
        prev_node = None
        while current:
            if current.data == prev_node_data:
                prev_node = current
                break
            current = current.next

        if prev_node is None:
            print(f"Node with data {prev_node_data} not found.")
            return

        new_node = Node(data)
        new_node.next = prev_node.next
        prev_node.next = new_node

    # Method to delete a node by value
    def delete_node(self, key):
        current = self.head

        # If head node itself holds the key to be deleted
        if current and current.data == key:
            self.head = current.next
            current = None
            print(f"Node with data {key} deleted.")
            return

        prev = None
        while current and current.data != key:
            prev = current
            current = current.next

        # If key was not present in linked list
        if current is None:
            print(f"Node with data {key} not found.")
            return

        # Unlink the node from linked list
        prev.next = current.next
        current = None
        print(f"Node with data {key} deleted.")

    # Method to delete a node at a given position (index)
    def delete_at_position(self, position):
        if self.head is None:
            print("List is empty, nothing to delete.")
            return

        temp = self.head

        # If head needs to be removed
        if position == 0:
            self.head = temp.next
            temp = None
            print(f"Node at position {position} deleted.")
            return

        # Find previous node of the node to be deleted
        for i in range(position - 1):
            if temp is None or temp.next is None:
                print(f"Position {position} out of bounds.")
                return
            temp = temp.next

        # If position is more than number of nodes
        if temp is None or temp.next is None:
            print(f"Position {position} out of bounds.")
            return

        # Node temp.next is the node to be deleted
        next_node = temp.next.next
        temp.next = None
        temp.next = next_node
        print(f"Node at position {position} deleted.")

    # Method to search for a node with a given value
    def search(self, key):
        current = self.head
        position = 0
        while current:
            if current.data == key:
                print(f"Value {key} found at position {position}.")
                return True
            current = current.next
            position += 1
        print(f"Value {key} not found in the list.")
        return False


# Demonstration of Singly Linked List Operations
print("\nSingly Linked List Operations Demonstration ")

# Create an empty linked list
my_sll = SinglyLinkedList()

print("1. Initial List (empty):")
my_sll.print_list()

# Insert at the beginning
my_sll.insert_at_beginning(10)
my_sll.insert_at_beginning(5)
print("2. After inserting 5, 10 at beginning:")
my_sll.print_list()

# Insert at the end
my_sll.insert_at_end(20)
my_sll.insert_at_end(30)
print("3. After inserting 20, 30 at end:")
my_sll.print_list()

# Insert after a specific node
my_sll.insert_after(10, 15)
print("4. After inserting 15 after node with data 10:")
my_sll.print_list()
my_sll.insert_after(100, 99)

# Search for an element
print("5. Searching for 20:")
my_sll.search(20)
print("6. Searching for 100:")
my_sll.search(100)

# Delete a node by value
my_sll.delete_node(15)
print("7. After deleting node with data 15:")
my_sll.print_list()

# Delete a node by value (not found)
my_sll.delete_node(100)
print("8. After trying to delete node with data 100:")
my_sll.print_list()

# Delete a node at a specific position
my_sll.delete_at_position(0)
print("9. After deleting node at position 0:")
my_sll.print_list()

my_sll.delete_at_position(1)
print("10. After deleting node at position 1:")
my_sll.print_list()

my_sll.delete_at_position(5)
my_sll.print_list()
```

Implementation of Doubly-linked list

```python
class DoublyLinkedListNode:
    def __init__(self, data):
        self.data = data
        self.next = None  # Pointer to next node
        self.prev = None  # Pointer to previous node
```

```python
class DoublyLinkedList:
    def __init__(self):
        self.head = None  # Head of the list
        self.tail = None  # Tail of the list (optional, but useful for doubly linked lists)

    # Method to print the list from head to tail
    def print_forward(self):
        current = self.head
        elements = []
        while current:
            elements.append(current.data)
            current = current.next
        print("Forward: " + " <-> ".join(map(str, elements)) if elements else "Forward: List is empty")

    # Method to print the list from tail to head
    def print_backward(self):
        current = self.tail
        elements = []
        while current:
            elements.append(current.data)
            current = current.prev
        print("Backward: " + " <-> ".join(map(str, reversed(elements))) if elements else "Backward: List is empty")

    # Insert at the beginning of the list
    def insert_at_beginning(self, data):
        new_node = DoublyLinkedListNode(data)
        if self.head is None:
            self.head = new_node
            self.tail = new_node
        else:
            new_node.next = self.head
            self.head.prev = new_node
            self.head = new_node

    # Insert at the end of the list
    def insert_at_end(self, data):
        new_node = DoublyLinkedListNode(data)
        if self.tail is None:
            self.head = new_node
            self.tail = new_node
        else:
            new_node.prev = self.tail
            self.tail.next = new_node
            self.tail = new_node

    # Insert after a given node's data
    def insert_after(self, prev_node_data, data):
        current = self.head
        while current and current.data != prev_node_data:
            current = current.next

        if current is None:
            print(f"Node with data {prev_node_data} not found. Cannot insert after.")
            return

        new_node = DoublyLinkedListNode(data)
        new_node.next = current.next
        new_node.prev = current
        if current.next:
            current.next.prev = new_node
        current.next = new_node

        if new_node.next is None: # If inserted at the end
            self.tail = new_node

    # Delete a node by its data value
    def delete_node(self, key):
        current = self.head

        while current and current.data != key:
            current = current.next

        if current is None:
            print(f"Node with data {key} not found. Cannot delete.")
            return

        # If it's the head node
        if current.prev is None:
            self.head = current.next
            if self.head:
                self.head.prev = None
            else: # List became empty
                self.tail = None
        # If it's the tail node
        elif current.next is None:
            self.tail = current.prev
            self.tail.next = None
        # If it's a middle node
        else:
            current.prev.next = current.next
            current.next.prev = current.prev

        print(f"Node with data {key} deleted.")

    # Search for a node with a given value
    def search(self, key):
        current = self.head
        position = 0
        while current:
            if current.data == key:
                print(f"Value {key} found at position {position}.")
                return True
            current = current.next
            position += 1
        print(f"Value {key} not found in the list.")
        return False
```

```python
# Create an empty doubly linked list
my_dll = DoublyLinkedList()

print("1. Initial List (empty):")
my_dll.print_forward()
my_dll.print_backward()

# Insert at the beginning
my_dll.insert_at_beginning(10)
my_dll.insert_at_beginning(5)
print("\n2. After inserting 5, 10 at beginning:")
my_dll.print_forward() # Expected: 5 <-> 10
my_dll.print_backward() # Expected: 10 <-> 5

# Insert at the end
my_dll.insert_at_end(20)
my_dll.insert_at_end(30)
print("\n3. After inserting 20, 30 at end:")
my_dll.print_forward() # Expected: 5 <-> 10 <-> 20 <-> 30
my_dll.print_backward() # Expected: 30 <-> 20 <-> 10 <-> 5

# Insert after a specific node
my_dll.insert_after(10, 15)
print("\n4. After inserting 15 after node with data 10:")
my_dll.print_forward() # Expected: 5 <-> 10 <-> 15 <-> 20 <-> 30
my_dll.print_backward() # Expected: 30 <-> 20 <-> 15 <-> 10 <-> 5
my_dll.insert_after(100, 99) # Test non-existent previous node

# Search for an element
print("\n5. Searching for 20:")
my_dll.search(20) # Expected: Found
print("6. Searching for 100:")
my_dll.search(100) # Expected: Not found

# Delete a node by value
my_dll.delete_node(15)
print("\n7. After deleting node with data 15:")
my_dll.print_forward() # Expected: 5 <-> 10 <-> 20 <-> 30
my_dll.print_backward() # Expected: 30 <-> 20 <-> 10 <-> 5

# Delete a node by value (head)
my_dll.delete_node(5)
print("\n8. After deleting head node (5):")
my_dll.print_forward() # Expected: 10 <-> 20 <-> 30
my_dll.print_backward() # Expected: 30 <-> 20 <-> 10

# Delete a node by value (tail)
my_dll.delete_node(30)
print("\n9. After deleting tail node (30):")
my_dll.print_forward() # Expected: 10 <-> 20
my_dll.print_backward() # Expected: 20 <-> 10

# Delete a node by value (not found)
my_dll.delete_node(100)
print("\n10. After trying to delete node with data 100:")
my_dll.print_forward()
my_dll.print_backward()

# Delete remaining nodes
my_dll.delete_node(10)
my_dll.delete_node(20)
print("\n11. After deleting all remaining nodes:")
my_dll.print_forward()
my_dll.print_backward()
my_dll.delete_node(50) # Delete from empty list
```
