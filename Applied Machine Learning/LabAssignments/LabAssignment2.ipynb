```python
#problem 7
"""
Find:

	Students scoring above 80.
	Students scoring between 75 and 90.
	Students belonging to CSE.
	CSE students scoring above 80.
"""
print("Students scoring above 80: \n",df[df['Marks']>80])
print("Students scoring between 75 and 90: \n",df[(df['Marks']>=75) & (df['Marks']<=90)])
print("Students belonging to CSE: \n",df[df['Department']=='CSE'])
print("CSE students scoring above 80: \n",df[(df['Department']=='CSE') & (df['Marks']>80)])
```

```python
#problem 9
"""
Add a new column called Result. Students scoring 75 or above should receive "Pass", otherwise "Fail".
"""
df['Result'] = np.where(df['Marks'] >= 75, 'Pass', 'Fail')
print(df)
```

```python
#problem 10
"""
Assign grades according to:

	| Marks | Grade |
	| ----: | ----- |
	|  ≥ 90 | A     |
	|  ≥ 80 | B     |
	|  ≥ 70 | C     |
	|  < 70 | D     |
"""

conditions = [
    (df['Marks'] >= 90),
    (df['Marks'] >= 80) & (df['Marks'] < 90),
    (df['Marks'] >= 70) & (df['Marks'] < 80),
    (df['Marks'] < 70)
]
choices = ['A', 'B', 'C', 'D']
df['Grade'] = np.select(conditions, choices, default='Unknown')

print(df)
```

```python
#problem 11
"""
Sort students:

	By marks in ascending order.
	By marks in descending order.
	By Department and then Marks.
"""
print("By marks in ascending order: \n",df.sort_values(by='Marks'))
print("By marks in descending order: \n",df.sort_values(by='Marks',ascending=False))
print("By Department and then Marks: \n",df.sort_values(by=['Department','Marks']))
```

```python
#Problem 17
"""
Consider:

data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE"
    ],

    "Marks": [
        85, 78, 92,
        74, 88, 80
    ]
}

df = pd.DataFrame(data)


Find:
Average marks by department
Maximum marks by department
Number of students in each department

For each department, calculate:
Mean
Minimum
Maximum
Count
"""
data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE"
    ],

    "Marks": [
        85, 78, 92,
        74, 88, 80
    ]
}

df = pd.DataFrame(data)
print("Average marks by department: \n",df.groupby('Department')['Marks'].mean())
print("-"*40)
print("Maximum marks by department: \n",df.groupby('Department')['Marks'].max())
print("-"*40)
print("Number of students in each department: \n",df['Department'].value_counts())
print("-"*40)
print("For each department, calculate: \n")
print("Mean: \n",df.groupby('Department')['Marks'].mean())
print("-"*30)
print("Minimum: \n",df.groupby('Department')['Marks'].min())
print("-"*30)
print("Maximum: \n",df.groupby('Department')['Marks'].max())
print("-"*30)
print("Count: \n",df.groupby('Department')['Marks'].count())
```

```python
#Problem 19
"""
Perform the following analysis:

Display the first five records.
Find the dimensions of the dataset.
Find the average final mark.
Find the highest-scoring student.
Find the lowest-scoring student.
Display students scoring above 80.
Find average marks by department.
Find students having attendance above 80%.
Sort students by FinalMarks.
Add a Result column: Pass if FinalMarks ≥ 70, otherwise Fail.
Add a performance category:
	≥90 → Excellent
	≥80 → Very Good
	≥70 → Good
	<70 → Needs Improvement
Find the department having the highest average FinalMarks.

"""
data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya",
        "John", "Priya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE",
        "ME", "ECE"
    ],

    "Attendance": [
        92, 75, 96, 68,
        82, 88, 72, 95
    ],

    "Internal": [
        85, 72, 94, 65,
        80, 86, 70, 91
    ],

    "FinalMarks": [
        88, 74, 96, 62,
        83, 89, 68, 93
    ]
}

df = pd.DataFrame(data)
print("First 5 records:")
print(df.head(5))
print("Dimensions of the dataset: ",df.shape)
print("Average final mark: ",df['FinalMarks'].mean())
print("Highest-scoring student: \n",df[df['FinalMarks'] == df['FinalMarks'].max()])
print("-"*60)
print("Lowest-scoring student: \n",df[df['FinalMarks'] == df['FinalMarks'].min()])
print("-"*60)
print("Students scoring above 80: \n",df[df['FinalMarks'] > 80])
print("-"*60)
print("Average marks by department: \n",df.groupby('Department')['FinalMarks'].mean())
print("-"*60)
print("Students having attendance above 80%: \n",df[df['Attendance'] > 80])
print("-"*60)
print("Sorted students by FinalMarks: \n",df.sort_values(by='FinalMarks'))
print("-"*60)
df['Result'] = np.where(df['FinalMarks'] >= 70, 'Pass', 'Fail')
print("Result column: \n",df)
print("-"*60)
df['Performance'] = np
np.where(df['FinalMarks'] >= 90, 'Excellent',
         np.where(df['FinalMarks'] >= 80, 'Very Good',
                  np.where(df['FinalMarks'] >= 70, 'Good', 'Needs Improvement')))
print("Department with the highest average FinalMarks: \n",df.groupby('Department')['FinalMarks'].mean().idxmax())
```

```python
#Problem 20
"""
Average salary.
Maximum and minimum salary.
Employee receiving the highest salary.
Employees earning above ₹60,000.
Average salary for each department.
Average experience by department.
Number of employees in each department.
Sort employees by salary.
Find IT employees with more than 3 years of experience.
Add a SalaryLevel column:
	≥ ₹75,000 → High
	₹50,000–₹74,999 → Medium
	< ₹50,000 → Low
"""
employees = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",
        "E05", "E06", "E07", "E08"
    ],

    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],

    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5
    ],

    "Salary": [
        60000, 45000, 75000, 65000,
        55000, 85000, 48000, 70000
    ]
})
print("Average salary: ",employees['Salary'].mean())
print("Maximum salary: ",employees['Salary'].max())
print("Minimum salary: ",employees['Salary'].min())
print("Employee receiving the highest salary: \n",employees[employees['Salary'] == employees['Salary'].max()])
print("-"*60)
print("Employee earning above 60,000: \n", employees[employees['Salary'] > 60000])
print("-"*60)
print("Average salary for each department: \n",employees.groupby('Department')['Salary'].mean())
print("-"*60)
print("Average experience by department: \n",employees.groupby('Department')['Experience'].mean())
print("-"*60)
print("Number of employees in each department: \n",employees['Department'].value_counts())
print("-"*60)
print("Sorted employees by salary: \n",employees.sort_values(by='Salary'))
print("-"*60)
print("IT employees with more than 3 years of experience: \n",employees[(employees['Department'] == 'IT') & (employees['Experience'] > 3)])
print("-"*60)
employees['SalaryLevel'] = np.where(employees['Salary'] >= 75000, 'High',
                                    np.where(employees['Salary'] >= 50000, 'Medium', 'Low'))
print("Employees with salary levels: \n",employees)
```
