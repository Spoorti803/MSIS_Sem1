Diabetics datas

```python
import pandas as pd
import numpy as np
df = pd.read_csv('diabetes.csv')
df.head(5)
```

```python
df['Pregnancies'].max()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
X = df.drop('Outcome',axis=1)
y = df['Outcome']
```

```python
np.array(X)
```

```python
print(X.shape)
print(y.shape)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_sc = sc.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X_sc,y,test_size=0.35,random_state=10)
```

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier()
dt.fit(X_train,y_train)
```

```python
y_pred = dt.predict(X_test)
y_pred
```

```python
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
print(classification_report(y_test,y_pred))
accuracy = accuracy_score(y_test,y_pred)
print("Accuracy score:",accuracy*100)
confusion_matrix(y_test,y_pred)
```

```python
new_data = np.array([[  1.   ,  85.   ,  66.   ,  29.   ,   0.   ,  26.6  ,   0.351,  31.   ]])
new_data = sc.transform(new_data)
result = dt.predict(new_data)
if result == 1:
  print("Customer is diabetic")
else:
  print("Customer is not diabetic")
```

## **iris dataset**

```python
import pandas as pd
df = pd.read_csv('Iris.csv')
df.head(5)
```

```python
df.describe()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
X = df.drop('Species',axis=1)
y = df['Species']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.35,random_state=10)
```

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier()
dt.fit(X_train,y_train)
```

```python
y_pred = dt.predict(X_test)
y_pred
```

```python
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
print(classification_report(y_test,y_pred))
accuracy = accuracy_score(y_test,y_pred)
print("Accuracy score:",accuracy*100)
confusion_matrix(y_test,y_pred)
```
