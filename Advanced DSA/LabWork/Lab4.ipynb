```python
#implementing stacks = push, pop, peek

class stack:
  def __init__(self):
    self.items = []

  def push(self, item):
    self.items.append(item)

  def pop(self):
    return self.items.pop()

  def peek(self):
    return self.items

s = stack()
print("Initial Stack: ", s.items)
s.push(1)
s.push(2)
s.push(3)
s.push(4)
print("Stack after pushing the elements",s.items)
s.pop()
print("Stack after poping an element",s.items)
```

```python
#implementing queue = enqueue,deque,front ele(peek)
class queue:
  def __init__(self,q):
    self.q = q
  def enqueue(self,item):
    self.q.append(item)
  def deque(self):
    return self.q.pop(0)
  def front(self):
    return self.q[0]

def main():
  elements = [1,2,3]
  q = queue(elements)
  print("Queue before adding more elemnts",q.q)

  q.enqueue(4)
  print("queue after adding an element",q.q)
  q.deque()
  print("queue after deleting an element",q.q)
  print("First element of the queue",q.front())
if __name__ == "__main__":
  main()
```
