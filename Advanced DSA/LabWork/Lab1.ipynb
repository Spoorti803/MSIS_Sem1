List Implemenation

```python
my_array = [10, 20, 30, 40, 50]
print(f"Initial array: {my_array}")

# Insert Operation ---
# Insert at the end
my_array.append(60)
print(f"Array after appending 60: {my_array}")

# Insert at a specific index (e.g., insert 25 at index 2)
my_array.insert(2, 25)
print(f"Array after inserting 25 at index 2: {my_array}")

# --- Delete Operation ---
# Delete by value (e.g., remove 40)
if 40 in my_array:
    my_array.remove(40)
print(f"Array after removing 40: {my_array}")

# Delete by index (e.g., remove element at index 0)
deleted_element = my_array.pop(0)
print(f"Array after popping element at index 0 ({deleted_element}): {my_array}")

# --- Update Operation ---
# Update an element (e.g., change element at index 1 to 22)
if len(my_array) > 1:
    my_array[1] = 22
print(f"Array after updating element at index 1 to 22: {my_array}")

# --- Search Operation ---
# Search for a value (e.g., search for 22)
search_value = 22
if search_value in my_array:
    print(f"Value {search_value} found in array.")
    print(f"Index of {search_value}: {my_array.index(search_value)}")
else:
    print(f"Value {search_value} not found in array.")

# Search for a value that does not exist
search_value = 100
if search_value in my_array:
    print(f"Value {search_value} found in array.")
else:
    print(f"Value {search_value} not found in array.")
```
