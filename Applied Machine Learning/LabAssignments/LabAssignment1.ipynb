```python
#problem 23
#Create a NumPy array containing 100 random exam marks and determine how many students fall into the ranges 0–39, 40–59, 60–79, and 80–100.
arr = np.random.randint(1,100,100)
print("the number of students in the range of 0 to 39: ",len(arr[arr<40]))
print("the number of students in the range of 40 to 59: ",len(arr[(arr>40) & (arr<60)]))
print("the number of students in the range of 60 to 79: ",len(arr[(arr>60) & (arr<80)]))
print("the number of students in the range of 80 to 100: ",len(arr[arr>80]))
```

```python
#problem 24
#Compare the time required to multiply 1,000,000 numbers using a Python list and a NumPy array.


import numpy as np
import time

size = 1_000_000 # 1,000,000 numbers

# Create a Python list
python_list = list(range(size))

# Create a NumPy array
numpy_array = np.arange(size)

# Measure time for Python list multiplication
start_time = time.time()
python_result = [x * 2 for x in python_list]
end_time = time.time()
python_time = end_time - start_time
print(f"Time taken for Python list multiplication: {python_time:.6f} seconds")

# Measure time for NumPy array multiplication
start_time = time.time()
numpy_result = numpy_array * 2
end_time = time.time()
numpy_time = end_time - start_time
print(f"Time taken for NumPy array multiplication: {numpy_time:.6f} seconds")

print(f"\nNumPy is approximately {python_time / numpy_time:.2f} times faster than Python lists for this operation.")
```
