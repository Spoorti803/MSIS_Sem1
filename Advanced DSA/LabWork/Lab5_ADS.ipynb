```python
class Circular_queue:
  def __init__(self, capacity):
    self.capacity = capacity
    self.queue = [None] * capacity
    self.front = -1
    self.rear = -1
    self.size = 0

  def is_empty(self):
    return self.size == 0

  def is_full(self):
    return self.size == self.capacity

  def enqueue(self, ele):
    if self.is_full():
      print("The queue is full")
      return

    if self.is_empty():
      self.front = 0

    self.rear = (self.rear + 1) % self.capacity
    self.queue[self.rear] = ele
    self.size += 1
    print(f"Enqueued {ele}. Current queue: {self.queue}, Front: {self.front}, Rear: {self.rear}, Size: {self.size}")

  def deque(self):
    if self.is_empty():
      print("The queue is empty")
      return None

    removed_element = self.queue[self.front]
    self.queue[self.front] = None
    self.size -= 1

    if self.is_empty():
      self.front = -1
      self.rear = -1
    else:
      self.front = (self.front + 1) % self.capacity

    print(f"Dequeued {removed_element}. Current queue: {self.queue}, Front: {self.front}, Rear: {self.rear}, Size: {self.size}")
    return removed_element


def main():
  print("Demonstrating Circular Queue")

  cque = Circular_queue(4)

  cque.enqueue(10)
  cque.enqueue(30)
  cque.enqueue(40)

  cque.enqueue(50)

  dequeued_item = cque.deque()
  print(f"Returned by deque: {dequeued_item}")

  cque.enqueue(60)

  dequeued_item = cque.deque()
  print(f"Returned by deque: {dequeued_item}")

  cque.deque()
  cque.deque()
  cque.deque()

  cque.deque() # Attempt to deque when empty

if __name__ == "__main__":
  main()
```
